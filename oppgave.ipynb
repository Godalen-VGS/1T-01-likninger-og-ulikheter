## Rasjonale likninger

Når $x$ står i nevneren, må vi passe på at nevneren ikke blir null. Sympy tar hensyn til dette og tar ikke med «falske» løsninger.

```python
import sympy as sp
x = sp.symbols('x')

likning = sp.Eq(x/(x - 2), 2/(x - 2) + 3)
print(sp.solve(likning, x))
```

Ganger du med $x - 2$ på begge sider, får du $x = 2 + 3(x - 2)$, som gir $x = 2$. Men sympy gir en tom liste. Hvorfor? Sett inn $x = 2$ i den opprinnelige likningen og se hva som skjer.

## Ulikheter

For ulikheter bruker vi `sp.solveset(ulikhet, x, domain=sp.S.Reals)`. Svaret blir et **intervall**:

| Sympy skriver | Betyr | Vanlig skrivemåte |
| --- | --- | --- |
| `Interval(1, 4)` | $1 \le x \le 4$ | $[1, 4]$ |
| `Interval.open(1, 4)` | $1 < x < 4$ | $\langle 1, 4 \rangle$ |
| `Interval(1, oo)` | $x \ge 1$ | $[1, \to\rangle$ |
| `Union(A, B)` | $x$ ligger i $A$ eller $B$ | $A \cup B$ |

```python
print(sp.solveset(3*x + 1 < 7, x, domain=sp.S.Reals))
print(sp.solveset(x**2 - 1 >= 0, x, domain=sp.S.Reals))
```

## Fortegnslinje som graf

En ulikhet som $f(x) > 0$ betyr: *hvor ligger grafen over $x$-aksen?* Grafen er derfor en god kontroll av svaret.

```python
import numpy as np
import matplotlib.pyplot as plt

xs = np.linspace(-3, 3, 300)
ys = xs**2 - 1

plt.plot(xs, ys)
plt.fill_between(xs, ys, 0, where=ys >= 0, alpha=0.3, label="$x^2 - 1 \\geq 0$")
plt.axhline(0, color="black", linewidth=0.8)
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Kjør denne cellen for å sjekke svarene dine. Du trenger ikke forstå koden.
def naer(a, b, rel=1e-3):
    """Sant hvis tallet a er innenfor rel (relativt) av b."""
    try:
        a, b = float(a), float(b)
    except (TypeError, ValueError):
        return False
    if b == 0:
        return abs(a) < 1e-9
    return abs(a - b) <= rel * abs(b)


def S(tekst):
    """Gjør tekst om til et sympy-uttrykk, f.eks. S("x**2 - 1")."""
    import sympy
    return sympy.sympify(tekst, locals={n: sympy.Symbol(n) for n in ("N", "O", "Q", "S")})


def lik(a, b):
    """Sant hvis a og b er matematisk like.

    Lister og mengder sammenlignes uten hensyn til rekkefølge, tupler (koordinater) i rekkefølge.
    """
    import sympy
    if a is None:
        return False
    if isinstance(b, tuple):
        if not isinstance(a, (tuple, list, sympy.Tuple)) or len(a) != len(b):
            return False
        return all(lik(x, y) for x, y in zip(a, b))
    if isinstance(b, (list, set)):
        if isinstance(a, sympy.FiniteSet):
            a = list(a)
        if not isinstance(a, (list, tuple, set)) or len(a) != len(b):
            return False
        rest = list(a)
        for y in b:
            for i, x in enumerate(rest):
                if lik(x, y):
                    del rest[i]
                    break
            else:
                return False
        return True
    if isinstance(b, dict):
        if not isinstance(a, dict) or len(a) != len(b):
            return False
        a = {str(k): v for k, v in a.items()}
        return all(str(k) in a and lik(a[str(k)], v) for k, v in b.items())
    try:
        a, b = sympy.sympify(a), sympy.sympify(b)
    except (sympy.SympifyError, TypeError):
        return False
    vanlige = lambda u: u.subs({s: sympy.Symbol(s.name) for s in u.free_symbols})
    a, b = vanlige(a), vanlige(b)
    if isinstance(a, sympy.Set) or isinstance(b, sympy.Set):
        return a == b
    if getattr(b, "is_Relational", False):
        return a == b
    try:
        return sympy.simplify(a - b) == 0
    except TypeError:
        return a == b

SJEKKER = [
    ('1a', 'losning_1a skal være løsningene av 2x² - 3x - 2 = 0', "lik(losning_1a, [S('-1/2'), 2])"),
    ('1b', 'losning_1b skal være løsningene av (x + 3)/(x - 1) = x', 'lik(losning_1b, [-1, 3])'),
    ('1c', 'losning_1c skal være løsningene av x³ = 4x', 'lik(losning_1c, [-2, 0, 2])'),
    ('2a', 'ulikhet_2a skal være intervallet x > 4', "lik(ulikhet_2a, S('Interval.open(4, oo)'))"),
    ('2b', 'ulikhet_2b skal være intervallet -2 ≤ x ≤ 3', "lik(ulikhet_2b, S('Interval(-2, 3)'))"),
    ('2c', 'ulikhet_2c skal være x < -2 eller x > 2', "lik(ulikhet_2c, S('Union(Interval.open(-oo, -2), Interval.open(2, oo))'))"),
    ('3', 'ulikhet_3 skal være x < -1 eller x ≥ 2', "lik(ulikhet_3, S('Union(Interval.open(-oo, -1), Interval(2, oo))'))"),
    ('4', 'bredder skal være intervallet 5 ≤ x ≤ 10', "lik(bredder, S('Interval(5, 10)'))"),
    ('5', 'k_to_losninger skal være k < -2 eller k > 6', "lik(k_to_losninger, S('Union(Interval.open(-oo, -2), Interval.open(6, oo))'))"),
]


def test_alt():
    feil = 0
    for _, beskrivelse, uttrykk in SJEKKER:
        try:
            ok = eval(uttrykk)
        except Exception as e:
            ok = False
            beskrivelse += f" (krasjet: {type(e).__name__}: {e})"
        if not ok:
            print("❌", beskrivelse)
            feil += 1
    if feil == 0:
        print(f"✅ Alle testene bestått! ({len(SJEKKER)} av {len(SJEKKER)})")
    else:
        print(f"\n{len(SJEKKER) - feil} av {len(SJEKKER)} tester bestått.")


test_alt()
```
